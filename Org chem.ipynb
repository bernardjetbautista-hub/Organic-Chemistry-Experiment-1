```python
def run_section(title: str, items: list) -> None:
    print(f"\n=== {title} ===")
    for item in items:
        print(item)
        input("\n--- Press [Enter] to continue ---\n")

def process_step(step, pause=False):
    """Prints step text and safely displays associated image if valid."""
    print(step["text"])

    # Safely handle None values, missing keys, and whitespace
    image_path = (step.get("image") or "").strip()

    if image_path and image_path.lower() != "none":
        if os.path.exists(image_path):
            display(Image(filename=image_path, width=400))
        else:
            print(f"⚠️ Image not found: {image_path}")

    if pause:
        input("\n--- Press [Enter] to continue ---\n")
```
